```python
#IQR table: checking how many outliers there are per feature

# Select numerical features (exclude target) 


numeric_cols = df.select_dtypes(include="number").drop(
    columns="Gallstone Status"
)

# Calculate IQR boundaries
Q1 = numeric_cols.quantile(0.25)
Q3 = numeric_cols.quantile(0.75)
IQR = Q3 - Q1

# Create outlier summary table
outlier_df = pd.DataFrame({
    "Lower Bound": Q1 - 1.5 * IQR,
    "Upper Bound": Q3 + 1.5 * IQR,
    "Outlier Count": (
        (numeric_cols < Q1 - 1.5 * IQR) |
        (numeric_cols > Q3 + 1.5 * IQR)
    ).sum()
}).round(2)

# Sort and display
outlier_df = outlier_df.sort_values(
    "Outlier Count", ascending=False
)

print(outlier_df)
```
