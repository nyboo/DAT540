```python
#VISUALIZATION

import matplotlib.pyplot as plt
import seaborn as sns

numeric_cols = df.select_dtypes(include='number').columns
#separating the binary value columns and the continuous value columns

binary_cols = [
    col for col in numeric_cols
    if df[col].dropna().nunique() == 2
]

continuous_cols = [
    col for col in numeric_cols
    if col not in binary_cols and col != "Gallstone Status"
]

#HISTOGRAMS
df[continuous_cols].hist(
    figsize = (16, 20),
    bins = 30,
    edgecolor = "black",
    layout = (10, 4)
)

plt.suptitle("Distribution of Continuous Numerical features", fontsize = 16)
plt.tight_layout()
plt.show()

#BOXPLOTS

plt.figure(figsize = (14, 12))

sns.boxplot(data =df[continuous_cols], orient = "h")

plt.title("Boxplots of continuous numerical features")
plt.tight_layout()
plt.show()

#BAR CHARTS FOR BINARY FEATURES

df[binary_cols].hist(
    figsize = (14,8),
    bins = [-0.5, 0.5, 1.5],
    edgecolor = "black",
    layout = (4, 4)
)

plt.suptitle("Distribution of binary features", fontsize = 16)
plt.tight_layout()
plt.show()


#CORRELATION HEATMAP

plt.figure(figsize = (16, 12))

sns.heatmap(
    df[numeric_cols.drop("Gallstone Status", errors = "ignore)")].corr(), 
    cmap = "coolwarm",
    center = 0
)

plt.title("Correlation between Numerical features")
plt.tight_layout()
plt.show()

```

```python
#IQR table: checking how many outliers there are per feature

# Select numerical features (exclude target) 


numeric_cols = df.select_dtypes(include="number").drop(
    columns="Gallstone Status"
)

# Calculate IQR boundaries
Q1 = numeric_cols.quantile(0.25)
Q3 = numeric_cols.quantile(0.75)
IQR = Q3 - Q1

# Create outlier summary table
outlier_df = pd.DataFrame({
    "Lower Bound": Q1 - 1.5 * IQR,
    "Upper Bound": Q3 + 1.5 * IQR,
    "Outlier Count": (
        (numeric_cols < Q1 - 1.5 * IQR) |
        (numeric_cols > Q3 + 1.5 * IQR)
    ).sum()
}).round(2)

# Sort and display
outlier_df = outlier_df.sort_values(
    "Outlier Count", ascending=False
)

print(outlier_df)
```
