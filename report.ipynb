```python
#IQR OUTLIER ANALYSIS
#using only continuous features
#to identify outliers

continuous_data = df[continuous_cols]

#calculate Q1, Q3 and IQR
Q1 = continuous_data.quantile(0.25)
Q3 = continuous_data.quantile(0.75)
IQR = Q3 - Q1

#outlier if you are outside lower bound or upper bound
outlier_df = pd.DataFrame({
    "Lower Bound": Q1 - 1.5 * IQR,
    "Upper Bound": Q3 + 1.5 * IQR,
    "Outlier Count": (
        (continuous_data < Q1 - 1.5 * IQR) |
        (continuous_data > Q3 + 1.5 * IQR)
    ).sum()
}).round(2)

print(outlier_df.sort_values("Outlier Count", ascending=False))


#Set the inconsistent Obesity (%) value at index 239 to NaN
df.loc[239, "Obesity (%)"] = float("nan")
```

```python
# FEATURES AND TARGET
# X will be all the predictor features (excluding the target feature)
X = df.drop(columns="Gallstone Status")
# y will be the target feature (Gallstone Status) which is the feature we want to predict
y = df["Gallstone Status"]

print(X.shape)  
print(y.shape)
```

```python
#TRAIN & TEST SPLIT
#splitting the data into one part we use for training and another for testing the model
#we will use a 80/20 split for training and testing
import sklearn as skl
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, 
    y, 
    test_size=0.2, 
    random_state=13, 
    stratify=y)

#test size --> 20% of the data will be split into testing
#random state --> makes sure the computer splits the data the same way every time??
#stratify = Y --> makes sure the ratio of the target stays the same in both the training and testing data

print("Training:", X_train.shape, y_train.shape)
print("Testing:", X_test.shape, y_test.shape)
```

```python
# PREPROCESSING

# change NaN value in the obesity column to the median of the test data
from sklearn.impute import SimpleImputer

# choose median as the value used for the missing data
imputer = SimpleImputer(strategy="median")

# find the medians from the training data and replaces missing values
# uses the same median for testing data - does not recalculate it
# ravel gets it in the correct format
X_train["Obesity (%)"] = imputer.fit_transform(X_train[["Obesity (%)"]]).ravel()
X_test["Obesity (%)"] = imputer.transform(X_test[["Obesity (%)"]]).ravel()

```

```python
# FEATURE SELECTION
# finding the features that are most useful for predicting the target

#check correlation between each feature and the gallstone status (target feature)
#corrwith(y_train) finds the correlation of each feature with y (gallstone status)
target_corr = X_train.corrwith(y_train).abs().sort_values(ascending=False)
print(target_corr)

#find features that are strongly related to each other, ex. weight, height and BMI
import numpy as np

# calculate correlation between all features
corr = X_train.corr().abs()

#keep each pair only once
high_corr = corr.where(
    np.triu(np.ones(corr.shape), k=1).astype(bool)
    ).stack().sort_values(ascending=False)

#show the feature pairs with correlation above 0.8

print(high_corr[high_corr > 0.8])

# remove some of the features that contain similar information to other features

features_to_remove = [
    "Total Body Fat Ratio (TBFR) (%)",
    "Obesity (%)",
    "Total Fat Content (TFC)",
    "Muscle Mass (MM)",
    "Total Cholesterol (TC)",
    "Total Body Water (TBW)",
    "Visceral Fat Rating (VFR)"
]
#TBFR is highly correlated with lean mass, but lean mass has a slightly higher correlation with gallstone status
#obesity is highly correlated with BMI, but BMI has a higher correlation with gallstone status
#total fat content is highly correlated with several body composition features, so it was removed to reduce repeated information
#muscle mass is highly correlated with several body composition features, so it was removed to reduce repeated information
#total cholesterol is highly correlated with LDL, but LDL has a higher correlation with gallstone status
#total body water is highly correlated with extracellular water, but extracellular water has a higher correlation with gallstone status
#VFR is highly correlated with total fat content and weight, so it was removed to reduce repeated information

#after having these features removed there are only a small number of correlations left above 0.8
#these are kept becuse they were close to the threshold of 0.8

#Make a new list with the selected features where we drop the list we defined above

X_train_selected = X_train.drop(columns = features_to_remove)
X_test_selected = X_test.drop(columns = features_to_remove)

#Print to check that the features have been removed
print("Before, training:", X_train.shape)
print("After, training:", X_train_selected.shape)
print("Before, testing:", X_test.shape)
print("After, testing:", X_test_selected.shape)

```

```python
# MODEL SELECTION
#Research question - forslag under:
#Which machine learning model performs best in predicting gallstone status based on the selected features?

#forslag til modellar:
    #logisitic regression - simple classification model for predicting binary target
    #decision tree - uses decision rules to classify patients and find non-linear relationships in the data
    #random forest - combines many decision trees to make more reliable predicitons and reduce overfitting


#importing the models from the sklearn library (they have standard tools for these models)
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# import pipeline so that we can scale the data automatically inside the model

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# random_state makes the random parts of the model repeatable so we get the same results each time
# 13 is an arbitrary number chosen to make the random parts of the model repeatable (it is my favourit number:)
# the pipeline ensures that the data is scaled before being passed to the logistic regression model
log_model = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=1000))
])

tree_model = DecisionTreeClassifier(random_state=13)
forest_model = RandomForestClassifier(random_state=13)


#MODEL TRAINING
#fit lets each model learn the relationship between the features 
#and the gallstone status

# logistic regression uses the pipeline model so that the data is scaled automatically by the pipeline model
log_model = log_model.fit(X_train_selected, y_train)

#decision tree and random forest do not required scaled data
#decision tree uses splitting rules
#random forest is many decision trees
tree_model.fit(X_train_selected, y_train)
forest_model.fit(X_train_selected, y_train)



# TRAINING PERFORMANCE
#let each trained model make predictions on the training data
log_train_pred = log_model.predict(X_train_selected)

tree_train_pred = tree_model.predict(X_train_selected)

forest_train_pred = forest_model.predict(X_train_selected)
#we are now making each model predict the gallstone status for the training data

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
#using sklearn because this library provied both functions for the models
#and because they provide functions for evaluating model performance

#accuracy score checks how many predictions were correct overall
#precision score checks how many of the predicted positives were actually positive
#recall score checks how many of the actual positives were correctly predicted
#f1 score is the mean of precision and recall

models = {
    "Logistic Regression": log_train_pred,
    "Decision Tree": tree_train_pred,
    "Random Forest": forest_train_pred
}

for name, predictions in models.items():
    print(name)
    print("Accuracy:", round(accuracy_score(y_train, predictions), 3))
    print("Precision:", round(precision_score(y_train, predictions, pos_label=0), 3))
    print("Recall:", round(recall_score(y_train, predictions, pos_label=0), 3))
    print("F1 Score:", round(f1_score(y_train, predictions, pos_label=0), 3))
    print()

#pos_label=0 because we want ot evaluate how good it is at finding the cases with gallstones which in this dataset is 0
# basically gallstones present = 0 and no gallstones = 1 


# CROSS VALIDATION
# check how the models perform on different parts of the training data

from sklearn.model_selection import cross_val_score

log_cv = cross_val_score(
    log_model, X_train_selected, y_train, cv=5,
scoring="accuracy"
)


tree_cv = cross_val_score(
    tree_model, X_train_selected, y_train, cv=5,
    scoring="accuracy"
)

forest_cv = cross_val_score(
    forest_model, X_train_selected, y_train, cv=5,
    scoring="accuracy"
)

print ("Logistic regression:", log_cv, "Mean:", log_cv.mean().round(3))
print ("Decision Tree:", tree_cv, "Mean:", tree_cv.mean().round(3))
print ("Random Forest:", forest_cv, "Mean:", forest_cv.mean().round(3))

# Logistic Regression had the highest cross-validation accuracy (81.6%)
# Random Forest performed similarly with 80.4%
# Decision Tree dropped from 100% training accuracy to 71.4% in cross-validation,
# which suggests that the model is overfitting the training data


#PARAMETER TUNING
# GridSearchCV tests different model settings using cross validation to find the best combination of hyperparameters?

from sklearn.model_selection import GridSearchCV

#tuning the decision tree parameters using GridSearchCV

tree_params = {
    "max_depth": [3, 5, 7, 10, None],
    "min_samples_split": [2, 5, 10]
}

tree_grid = GridSearchCV(
    DecisionTreeClassifier(random_state = 13),
    tree_params,
    cv=5,
    scoring="accuracy"
)

tree_grid.fit(X_train_selected, y_train)

print("Best settings for decision tree:", tree_grid.best_params_)
print("Best CV accuracy for decision tree:", round(tree_grid.best_score_, 3))


# tuning the random forest parameters using GridSearchCV

forest_params = {
    "n_estimators": [50, 100, 200],
    "max_depth": [3, 5, 10, None],
    "min_samples_split": [2, 5, 10]
}

forest_grid = GridSearchCV(
    RandomForestClassifier(random_state=13),
    forest_params,
    cv=5,
    scoring="accuracy"
)

forest_grid.fit(X_train_selected, y_train)

print("Best settings for random forest:", forest_grid.best_params_)
print("Best CV accuracy for random forest:", round(forest_grid.best_score_, 3))

# tuning the logistic regression

log_params = {
    "model__C": [0.01, 0.1, 1, 10, 100]
}

log_grid = GridSearchCV(
    log_model,
    log_params,
    cv=5,
    scoring="accuracy"
)

log_grid.fit(X_train_selected, y_train)

print("Best settings for logistic regression:", log_grid.best_params_)
print("Best CV accuracy for logistic regression:", round(log_grid.best_score_, 3))

```

```python
# OUR BEST MODELS
best_log = log_grid.best_estimator_
best_tree = tree_grid.best_estimator_
best_forest = forest_grid.best_estimator_
# this function finds the best estimator from the grid search for each model

# TESTING
# see how the tuned models predict gallstone status on unseen test data
log_test_pred = best_log.predict(X_test_selected)
tree_test_pred = best_tree.predict(X_test_selected)
forest_test_pred = best_forest.predict(X_test_selected)


#copy code from the training performance evaluation section, but for the test data
test_models = {
    "Logistic Regression": log_test_pred,
    "Decision Tree": tree_test_pred,
    "Random Forest": forest_test_pred
}

for name, predictions in test_models.items():
    print(name)
    print("Accuracy:", round(accuracy_score(y_test, predictions), 3))
    print("Precision:", round(precision_score(y_test, predictions, pos_label=0), 3))
    print("Recall:", round(recall_score(y_test, predictions, pos_label=0), 3))
    print("F1 Score:", round(f1_score(y_test, predictions, pos_label=0), 3))
    print()
```

```python
#FINAL MODEL COMPARISON

# Logistic Regression had the highest test accuracy and F1 score, meaning it had the best overall performance and balance.
# Decision Tree had the highest recall, meaning it detected the most actual gallstone cases. 
# Random Forest had the highest precision, meaning its predicted gallstone cases were most often correct.
# Overall, logistic regression performed best, while decision tree was best at identifying actual gallstone cases

# TABLE WITH FINAL MODEL RESULTS

results = pd.DataFrame({
    "Model": ["Logistic Regression", "Decision Tree", "Random Forest"],
    "Accuracy": [
        accuracy_score(y_test, log_test_pred),
        accuracy_score(y_test, tree_test_pred),
        accuracy_score(y_test, forest_test_pred)
    ],
    "Precision": [
        precision_score(y_test, log_test_pred, pos_label=0),
        precision_score(y_test, tree_test_pred, pos_label=0),
        precision_score(y_test, forest_test_pred, pos_label=0)
    ],
    "Recall": [
        recall_score(y_test, log_test_pred, pos_label=0),
        recall_score(y_test, tree_test_pred, pos_label=0),
        recall_score(y_test, forest_test_pred, pos_label=0  )
    ],
    "F1 Score": [
        f1_score(y_test, log_test_pred, pos_label=0),
        f1_score(y_test, tree_test_pred, pos_label=0),
        f1_score(y_test, forest_test_pred, pos_label=0)
    ]
})

# round the results to 3 decimals
results = results.round(3)

results
```
